```python
from tensorflow.keras.applications import MobileNetV2

model = MobileNetV2(weights="imagenet")

print("MobileNetV2 Model Loaded Successfully!")
```

```python
import tensorflow as tf

image_url = "https://storage.googleapis.com/download.tensorflow.org/example_images/YellowLabradorLooking_new.jpg"

image_path = tf.keras.utils.get_file(
    "dog.jpg",
    image_url
)

print("Image downloaded successfully!")
print(image_path)
```

```python
from tensorflow.keras.preprocessing import image
import numpy as np

img = image.load_img(image_path, target_size=(224, 224))

img_array = image.img_to_array(img)

img_array = np.expand_dims(img_array, axis=0)

print("Image prepared successfully!")
print("Image Shape:", img_array.shape)
```

```python
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input, decode_predictions

# Prepare image for MobileNetV2
processed_img = preprocess_input(img_array)

# Make prediction
predictions = model.predict(processed_img)

# Get top 3 predictions
results = decode_predictions(predictions, top=3)

# Display results
for result in results[0]:
    print("Label:", result[1])
    print("Confidence:", round(result[2] * 100, 2), "%")
    print()
```

```python
import matplotlib.pyplot as plt

plt.imshow(img)
plt.axis("off")
plt.title("Predicted: Labrador Retriever")
plt.show()
```

```python
image_url_2 = "https://images.unsplash.com/photo-1534528741775-53994a69daeb?w=600"

image_path_2 = tf.keras.utils.get_file(
    "person.jpg",
    image_url_2
)

print("Second image downloaded successfully!")
```

```python
img2 = image.load_img(image_path_2, target_size=(224, 224))

img_array2 = image.img_to_array(img2)
img_array2 = np.expand_dims(img_array2, axis=0)

processed_img2 = preprocess_input(img_array2)

predictions2 = model.predict(processed_img2)

results2 = decode_predictions(predictions2, top=3)

for result in results2[0]:
    print("Label:", result[1])
    print("Confidence:", round(result[2] * 100, 2), "%")
```

```python
print("IMAGE 1 - DOG")
print("Prediction:", results[0][0][1])
print("Confidence:", round(results[0][0][2] * 100, 2), "%")

print("\nIMAGE 2 - PERSON")
print("Prediction:", results2[0][0][1])
print("Confidence:", round(results2[0][0][2] * 100, 2), "%")
```

## How Does a Pretrained Model Work?

A pretrained model is a machine learning model that has already been trained on a large dataset.

In this project, I used MobileNetV2, a pretrained image classification model trained on the ImageNet dataset.

The model takes an image as input, processes it, and predicts the objects present in the image.

It provides:
- Predicted object labels.
- Confidence scores for each prediction.
- Top 3 possible classifications.

Using a pretrained model saves training time and allows us to perform image classification without training a model from scratch.
